# Dataset Simplificado

```python
import pandas as pd
import sqlite3
import re

# ==========================================
# 1. CARREGAR DATASET
# ==========================================

arquivo = "waves.csv"

df = pd.read_csv(arquivo)

print("Linhas:", len(df))
print("Colunas:", len(df.columns))


# ==========================================
# 2. PADRONIZAR NOMES DAS COLUNAS
# ==========================================

def normalize_column(col):

    col = col.strip().lower()
    col = re.sub(r"[^\w\s]", "", col)
    col = re.sub(r"\s+", "_", col)
    col = re.sub(r"_+", "_", col)

    return col

df.columns = [normalize_column(c) for c in df.columns]


# ==========================================
# 3. LIMPAR VALORES NULOS
# ==========================================

null_values = [
    "", "none", "null", "nan", "na", "n/a", "unknown"
]

df.replace(null_values, pd.NA, inplace=True)


# ==========================================
# 4. LIMPAR ESPAÇOS
# ==========================================

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()


# ==========================================
# 5. CRIAR DATA DO EVENTO
# ==========================================

if "anunciado_utc" in df.columns:

    df["data_evento"] = pd.to_datetime(
        df["anunciado_utc"], errors="coerce"
    )

elif "horario_provavel_de_lancamento" in df.columns:

    df["data_evento"] = pd.to_datetime(
        df["horario_provavel_de_lancamento"],
        errors="coerce"
    )

else:

    df["data_evento"] = pd.NaT


# ==========================================
# 6. PADRONIZAR COORDENADAS
# ==========================================

df["latitude"] = pd.to_numeric(
    df.get("target_lat"), errors="coerce"
)

df["longitude"] = pd.to_numeric(
    df.get("target_lon"), errors="coerce"
)


# ==========================================
# 7. IDENTIFICAR PAÍS
# ==========================================

if "paises_de_impacto" in df.columns:

    df["pais"] = df["paises_de_impacto"]

elif "landings_countries_iso" in df.columns:

    df["pais"] = df["landings_countries_iso"]

else:

    df["pais"] = None


# ==========================================
# 8. DATASET SIMPLES PARA ANÁLISE
# ==========================================

colunas_analise = [

    "id_da_onda",
    "operacao",
    "numero_da_onda",

    "data_evento",

    "pais",

    "latitude",
    "longitude",

    "drones_utilizados",
    "misseis_balisticos_utilizados",
    "misseis_de_cruzeiro_utilizados",

    "estimated_munitions_count",
    "estimated_intercept_rate",

    "danos",
    "mortes",
    "feridos",

    "interceptado"

]

df_analise = df[colunas_analise]

df_analise = df_analise.dropna(
    subset=["latitude", "longitude"]
)

df_analise.to_csv(
    "waves_dataset_geo_analise.csv",
    index=False
)

print("CSV de análise criado")


# ==========================================
# 9. CONSTRUIR DATASET CIENTÍFICO
# ==========================================

# ---------- EVENTS ----------

events = df[[
    "id_da_onda",
    "operacao",
    "numero_da_onda",
    "data_evento"
]].copy()

events.rename(columns={
    "id_da_onda": "event_id",
    "operacao": "operation",
    "numero_da_onda": "wave_number"
}, inplace=True)


# ---------- LOCATIONS ----------

locations = df[[
    "id_da_onda",
    "pais",
    "latitude",
    "longitude"
]].copy()

locations.rename(columns={
    "id_da_onda": "event_id",
    "pais": "country"
}, inplace=True)


# ---------- WEAPONS ----------

weapons = df[[
    "id_da_onda",
    "drones_utilizados",
    "misseis_balisticos_utilizados",
    "misseis_de_cruzeiro_utilizados",
    "estimated_munitions_count"
]].copy()

weapons.rename(columns={
    "id_da_onda": "event_id",
    "drones_utilizados": "drones",
    "misseis_balisticos_utilizados": "ballistic_missiles",
    "misseis_de_cruzeiro_utilizados": "cruise_missiles",
    "estimated_munitions_count": "total_munitions"
}, inplace=True)


# ---------- CASUALTIES ----------

casualties = df[[
    "id_da_onda",
    "mortes",
    "feridos"
]].copy()

casualties.rename(columns={
    "id_da_onda": "event_id",
    "mortes": "fatalities",
    "feridos": "injuries"
}, inplace=True)


# ---------- INTERCEPTIONS ----------

interceptions = df[[
    "id_da_onda",
    "interceptado",
    "estimated_intercept_rate"
]].copy()

interceptions.rename(columns={
    "id_da_onda": "event_id",
    "interceptado": "intercepted",
    "estimated_intercept_rate": "intercept_rate"
}, inplace=True)


# ==========================================
# 10. SALVAR BANCO CIENTÍFICO
# ==========================================

conn = sqlite3.connect(
    "geopolitics_conflict_dataset.sqlite"
)

events.to_sql("events", conn,
              if_exists="replace", index=False)

locations.to_sql("locations", conn,
                 if_exists="replace", index=False)

weapons.to_sql("weapons", conn,
               if_exists="replace", index=False)

casualties.to_sql("casualties", conn,
                  if_exists="replace", index=False)

interceptions.to_sql("interceptions", conn,
                     if_exists="replace", index=False)

conn.close()

print("Banco científico criado")
```
